Import modules

```python
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import VarianceThreshold
```

Load dataset penncath_final_balanced.csv

```python
dataset = pd.read_csv("../data/penncath_final_balanced.csv")
```

Split features (SNPs) and target variable (CAD)

```python
snp_cols = [col for col in dataset.columns if col != 'target']
X = dataset[snp_cols]
Y = dataset['target'].astype(int)
```

Handle missing values using most frequent genotype

```python
print("Missing values before: ", dataset.isnull().sum().sum())

imputer = SimpleImputer(strategy='most_frequent')
X_imputed = pd.DataFrame(imputer.fit_transform(X), columns=snp_cols, index=X.index)

print("Missing values after: ", X_imputed.isnull().sum().sum())
```

Convert SNPs to integer encoding (0,1,2)

```python
X_imputed = X_imputed.astype(int)
```

Remove low-variance SNPs

```python
selector = VarianceThreshold(threshold=0.01)
X_var = selector.fit_transform(X_imputed)
kept_cols = X_imputed.columns[selector.get_support()].tolist()
X_var = pd.DataFrame(X_var, columns=kept_cols, index=X_imputed.index)
```

Create final dataset

```python
df_clean = X_var.copy()
df_clean['target'] = Y
```

Save processed dataset to penncath_preprocessed.csv

```python
df_clean.to_csv('../data/penncath_preprocessed.csv', index=False)
```
